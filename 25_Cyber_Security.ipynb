```python
#TODO:-1.Password Strength Checker
import string
# ==========================================
# PASSWORD STRENGTH CHECKER
# ==========================================
def check_password(password):
    score = 0
    suggestions = []
    # --------------------------------------
    # 1. Check Length
    # --------------------------------------
    if len(password) >= 8:
        score += 1
    else:
        suggestions.append("Use at least 8 characters.")
    if len(password) >= 12:
        score += 1
    # --------------------------------------
    # 2. Check Lowercase
    # --------------------------------------
    if any(char.islower() for char in password):
        score += 1
    else:
        suggestions.append(
            "Add lowercase letters (a-z)."
        )
    # --------------------------------------
    # 3. Check Uppercase
    # --------------------------------------
    if any(char.isupper() for char in password):
        score += 1
    else:
        suggestions.append(
            "Add uppercase letters (A-Z)."
        )
    # --------------------------------------
    # 4. Check Digits
    # --------------------------------------
    if any(char.isdigit() for char in password):
        score += 1
    else:
        suggestions.append(
            "Add numbers (0-9)."
        )
    # --------------------------------------
    # 5. Check Special Characters
    # --------------------------------------
    if any(char in string.punctuation for char in password):
        score += 1
    else:
        suggestions.append(
            "Add special characters like @, #, $, %."
        )
    # --------------------------------------
    # 6. Check Common Passwords
    # --------------------------------------
    common_passwords = [
        "password",
        "123456",
        "12345678",
        "qwerty",
        "admin",
        "password123",
        "abc123",
        "letmein"
    ]
    if password.lower() in common_passwords:
        score = 0
        suggestions.append(
            "This is a commonly used password."
        )
    # --------------------------------------
    # 7. Determine Strength
    # --------------------------------------
    if score <= 2:
        strength = "WEAK"
    elif score <= 4:
        strength = "MEDIUM"
    elif score <= 5:
        strength = "STRONG"
    else:
        strength = "VERY STRONG"
    return score, strength, suggestions
# ==========================================
# MAIN PROGRAM
# ==========================================
def main():
    print("=" * 50)
    print("       PASSWORD STRENGTH CHECKER")
    print("=" * 50)
    password = input("\nEnter your password: ")
    score, strength, suggestions = check_password(
        password
    )
    print("\n" + "=" * 50)
    print("             RESULT")
    print("=" * 50)
    print(f"Password Length : {len(password)}")
    print(f"Score           : {score}/6")
    print(f"Strength        : {strength}")
    # --------------------------------------
    # Suggestions
    # --------------------------------------
    if suggestions:
        print("\nSuggestions:")
        for suggestion in suggestions:
            print("→", suggestion)
    else:
        print(
            "\nExcellent! Your password satisfies "
            "all basic checks."
        )
    print("=" * 50)
# ==========================================
# PROGRAM START
# ==========================================
if __name__ == "__main__":
    main()
```

```python
# TODO:-SECURE PASSWORD MANAGER
# │
# ├── Generate Password
# ├── Generate Multiple Passwords
# ├── Custom Character Selection
# │     ├── Uppercase
# │     ├── Lowercase
# │     ├── Numbers
# │     └── Symbols
# │
# ├── Password Strength
# │     ├── Length
# │     ├── Entropy
# │     └── Character diversity
# │
# ├── Copy Password
# ├── Save Password
# ├── Search Saved Password
# ├── Delete Password
# │
# ├── Master Password
# ├── Encryption
# ├── SQLite Database
# └── Tkinter GUI
#FIXME:-2.Password Generator
import string
import secrets
# ==========================================
# PASSWORD GENERATOR
# ==========================================
def generate_password(length):
    """
    Generates a secure random password.
    """
    # All possible characters
    characters = (
        string.ascii_letters +
        string.digits +
        string.punctuation
    )
    # Generate password
    password = ""
    for _ in range(length):
        password += secrets.choice(characters)
    return password
# ==========================================
# PASSWORD STRENGTH CHECKER
# ==========================================
def check_strength(password):
    score = 0
    if len(password) >= 8:
        score += 1
    if len(password) >= 12:
        score += 1
    if any(c.islower() for c in password):
        score += 1
    if any(c.isupper() for c in password):
        score += 1
    if any(c.isdigit() for c in password):
        score += 1
    if any(c in string.punctuation for c in password):
        score += 1
    if score <= 2:
        return "WEAK"
    elif score <= 4:
        return "MEDIUM"
    elif score == 5:
        return "STRONG"
    else:
        return "VERY STRONG"
# ==========================================
# DISPLAY PASSWORD
# ==========================================
def display_password(password):
    print("\n" + "=" * 50)
    print("GENERATED PASSWORD")
    print("=" * 50)
    print("Password :", password)
    print("Length   :", len(password))
    print("Strength :", check_strength(password))
    print("=" * 50)
# ==========================================
# MAIN PROGRAM
# ==========================================
def main():
    print("=" * 50)
    print("       🔐 SECURE PASSWORD GENERATOR")
    print("=" * 50)
    while True:
        print("\n1. Generate Password")
        print("2. Exit")
        choice = input("\nEnter your choice: ")
        # --------------------------------------
        # Generate password
        # --------------------------------------
        if choice == "1":
            try:
                length = int(
                    input(
                        "\nEnter password length: "
                    )
                )
                if length < 8:
                    print(
                        "\n❌ Password length should "
                        "be at least 8."
                    )
                    continue
                password = generate_password(length)
                display_password(password)
            except ValueError:
                print(
                    "\n❌ Please enter a valid number."
                )
        # --------------------------------------
        # Exit
        # --------------------------------------
        elif choice == "2":
            print(
                "\nThank you for using "
                "Password Generator!"
            )
            break
        else:
            print(
                "\n❌ Invalid choice!"
            )
# ==========================================
# PROGRAM START
# ==========================================
if __name__ == "__main__":
    main()
```

```python
#TODO:-3.File Encryption & Decryption Tool
# Encryption
      #        USER
      #          │
      #          ▼
      #     Password
      #          │
      #          ▼
      #   SHA-256 Hash
      #          │
      #          ▼
      #  Fernet Encryption Key
      #          │
      #          ▼
      #     Original File
      #          │
      #          ▼
      #     🔐 ENCRYPT
      #          │
      #          ▼
      # Encrypted File
# # Decryption
#       Encrypted File
#              │
#              ▼
#        User Password
#              │
#              ▼
#         SHA-256 Hash
#              │
#              ▼
#        Same Encryption Key
#              │
#              ▼
#           🔓 DECRYPT
#              │
#              ▼
#        Original File
import os
import base64
import hashlib
import getpass
from cryptography.fernet import Fernet
# ============================================================
# 1. GENERATE ENCRYPTION KEY FROM PASSWORD
# ============================================================
def generate_key(password):
    """
    Converts user password into a Fernet-compatible key.
    """
    password_bytes = password.encode()
    hash_value = hashlib.sha256(password_bytes).digest()
    key = base64.urlsafe_b64encode(hash_value)
    return key
# ============================================================
# 2. ENCRYPT FILE
# ============================================================
def encrypt_file(filename, password):
    try:
        # Check file exists
        if not os.path.exists(filename):
            print("\n❌ File does not exist.")
            return
        # Generate key
        key = generate_key(password)
        cipher = Fernet(key)
        # Read original file
        with open(filename, "rb") as file:
            data = file.read()
        # Encrypt data
        encrypted_data = cipher.encrypt(data)
        # Create encrypted filename
        encrypted_filename = filename + ".encrypted"
        # Save encrypted file
        with open(
            encrypted_filename,
            "wb"
        ) as file:
            file.write(encrypted_data)
        print("\n✅ File encrypted successfully!")
        print(
            "Encrypted file:",
            encrypted_filename
        )
    except Exception as e:
        print("\n❌ Encryption error:", e)
# ============================================================
# 3. DECRYPT FILE
# ============================================================
def decrypt_file(filename, password):
    try:
        # Check file exists
        if not os.path.exists(filename):
            print("\n❌ File does not exist.")
            return
        # Generate key
        key = generate_key(password)
        cipher = Fernet(key)
        # Read encrypted file
        with open(filename, "rb") as file:
            encrypted_data = file.read()
        # Decrypt
        decrypted_data = cipher.decrypt(
            encrypted_data
        )
        # Remove .encrypted
        if filename.endswith(".encrypted"):
            decrypted_filename = (
                filename[:-10]
            )
        else:
            decrypted_filename = (
                filename + ".decrypted"
            )
        # Save decrypted file
        with open(
            decrypted_filename,
            "wb"
        ) as file:
            file.write(decrypted_data)
        print("\n✅ File decrypted successfully!")
        print(
            "Decrypted file:",
            decrypted_filename
        )
    except Exception:
        print(
            "\n❌ Decryption failed!"
        )
        print(
            "Possible reasons:"
        )
        print(
            "1. Wrong password"
        )
        print(
            "2. File is corrupted"
        )
        print(
            "3. File was not encrypted by this tool"
        )
# ============================================================
# 4. SHOW FILE INFORMATION
# ============================================================
def file_info(filename):
    if not os.path.exists(filename):
        print("\n❌ File does not exist.")
        return
    size = os.path.getsize(filename)
    print("\n" + "=" * 50)
    print("FILE INFORMATION")
    print("=" * 50)
    print("File name :", filename)
    print("File size :", size, "bytes")
    print("=" * 50)
# ============================================================
# 5. MAIN MENU
# ============================================================
def main():
    while True:
        print("\n")
        print("=" * 55)
        print("🔐 FILE ENCRYPTION TOOL")
        print("=" * 55)
        print("1. Encrypt File")
        print("2. Decrypt File")
        print("3. File Information")
        print("4. Exit")
        print("=" * 55)
        choice = input(
            "Enter your choice: "
        )
        # ====================================================
        # ENCRYPT
        # ====================================================
        if choice == "1":
            filename = input(
                "\nEnter file path: "
            )
            password = getpass.getpass(
                "Enter encryption password: "
            )
            if len(password) < 8:
                print(
                    "\n❌ Password should contain "
                    "at least 8 characters."
                )
                continue
            encrypt_file(
                filename,
                password
            )
        # ====================================================
        # DECRYPT
        # ====================================================
        elif choice == "2":
            filename = input(
                "\nEnter encrypted file path: "
            )
            password = getpass.getpass(
                "Enter decryption password: "
            )
            decrypt_file(
                filename,
                password
            )
        # ====================================================
        # FILE INFORMATION
        # ====================================================
        elif choice == "3":
            filename = input(
                "\nEnter file path: "
            )
            file_info(filename)
        # ====================================================
        # EXIT
        # ====================================================
        elif choice == "4":
            print(
                "\nThank you for using "
                "File Encryption Tool!"
            )
            break
        else:
            print(
                "\n❌ Invalid choice."
            )
# ============================================================
# PROGRAM START
# ============================================================
if __name__ == "__main__":
    main()
#
# =======================================================
#        🔐 FILE ENCRYPTION TOOL
# =======================================================
# 1. Encrypt File
# 2. Decrypt File
# 3. File Information
# 4. Exit
# =======================================================
# Enter your choice: 1
#
# Enter file path: secret.txt
# Enter encryption password:
#
# ✅ File encrypted successfully!
# Encrypted file: secret.txt.encrypted
#
# Enter your choice: 2
#
# Enter encrypted file path: secret.txt.encrypted
# Enter decryption password:
#
# Enter your choice: 2
# Enter encrypted file path: secret.txt.encrypted
# Enter decryption password:
# ❌ Decryption failed!
# Possible reasons:
# 1. Wrong password
# 2. File is corrupted
# 3. File was not encrypted by this tool
```

```python
# FIXME:-4.Text Encryption App (Caesar Cipher)
# ============================================================
#              CAESAR CIPHER TEXT ENCRYPTION APP
# ============================================================

# ============================================================
# 1. ENCRYPT TEXT
# ============================================================

def encrypt_text(text, shift):
    encrypted = ""
    for char in text:
        # Uppercase letters
        if char.isupper():
            encrypted += chr(
                (ord(char) - ord('A') + shift) % 26
                + ord('A')
            )
        # Lowercase letters
        elif char.islower():
            encrypted += chr(
                (ord(char) - ord('a') + shift) % 26
                + ord('a')
            )
        # Numbers
        elif char.isdigit():
            encrypted += chr(
                (ord(char) - ord('0') + shift) % 10
                + ord('0')
            )
        # Spaces and special characters
        else:
            encrypted += char
    return encrypted
# ============================================================
# 2. DECRYPT TEXT
# ============================================================
def decrypt_text(text, shift):
    return encrypt_text(text, -shift)
# ============================================================
# 3. SHOW BANNER
# ============================================================
def show_banner():
    print("\n" + "=" * 60)
    print("        🔐 CAESAR CIPHER TEXT ENCRYPTION APP")
    print("=" * 60)
# ============================================================
# 4. MAIN PROGRAM
# ============================================================
def main():
    while True:
        show_banner()
        print("\n1. Encrypt Text")
        print("2. Decrypt Text")
        print("3. Exit")
        print("-" * 60)
        choice = input(
            "Enter your choice: "
        )
        # ====================================================
        # ENCRYPT
        # ====================================================
        if choice == "1":
            text = input(
                "\nEnter text to encrypt: "
            )
            try:
                shift = int(
                    input(
                        "Enter shift value: "
                    )
                )
                encrypted = encrypt_text(
                    text,
                    shift
                )
                print("\n" + "=" * 60)
                print("ENCRYPTION RESULT")
                print("=" * 60)
                print("Original Text :", text)
                print("Shift         :", shift)
                print("Encrypted Text:", encrypted)
            except ValueError:
                print(
                    "\n❌ Shift must be a number."
                )
        # ====================================================
        # DECRYPT
        # ====================================================
        elif choice == "2":
            text = input(
                "\nEnter encrypted text: "
            )
            try:
                shift = int(
                    input(
                        "Enter shift value: "
                    )
                )
                decrypted = decrypt_text(
                    text,
                    shift
                )
                print("\n" + "=" * 60)
                print("DECRYPTION RESULT")
                print("=" * 60)
                print("Encrypted Text:", text)
                print("Shift         :", shift)
                print("Decrypted Text:", decrypted)
            except ValueError:
                print(
                    "\n❌ Shift must be a number."
                )
        # ====================================================
        # EXIT
        # ====================================================
        elif choice == "3":
            print(
                "\n🔐 Thank you for using "
                "Caesar Cipher App!"
            )
            break
        else:

            print(
                "\n❌ Invalid choice."
            )
# ============================================================
# PROGRAM START
# ============================================================
if __name__ == "__main__":
    main()
#
# ============================================================
#         🔐 CAESAR CIPHER TEXT ENCRYPTION APP
# ============================================================
#
# 1. Encrypt Text
# 2. Decrypt Text
# 3. Exit
# ------------------------------------------------------------
# Enter your choice: 1
#
# Enter text to encrypt: Hello World
# Enter shift value: 3
#
# ============================================================
#               ENCRYPTION RESULT
# ============================================================
# Original Text : Hello World
# Shift         : 3
# Encrypted Text: Khoor Zruog
#
# ============================================================
#         🔐 CAESAR CIPHER TEXT ENCRYPTION APP
# ============================================================
#
# 1. Encrypt Text
# 2. Decrypt Text
# 3. Exit
# ------------------------------------------------------------
# Enter your choice: 2
#
# Enter encrypted text: Khoor Zruog
# Enter shift value: 3
#
# ============================================================
#               DECRYPTION RESULT
# ============================================================
# Encrypted Text: Khoor Zruog
# Shift         : 3
# Decrypted Text: Hello World
```

```python
# TODO:-4.Login Authentication System
# 1. Register User
# 2. Login
# 3. View Profile
# 4. Logout
# 5. Exit
#
# Security:
# ✓ Password hashing
# ✓ Random salt
# ✓ Password confirmation
# ✓ Failed login attempt limit
# ✓ Hidden password input
# ✓ Multiple users
#
# Python Login System
# │
# ├── Registration
# ├── Login
# ├── Password Hashing
# ├── SQLite Database
# ├── Forgot Password
# ├── Change Password
# ├── Email OTP
# ├── Account Lockout
# ├── Login History
# ├── Admin Panel
# ├── Role-Based Access
# ├── Session Management
# ├── JWT Authentication
# └── Tkinter GUI
import hashlib
import secrets
import getpass
# ============================================================
# DATABASE
# ============================================================
users = {}
# ============================================================
# PASSWORD HASHING
# ============================================================
def hash_password(password, salt=None):
    # Generate a random salt
    if salt is None:
        salt = secrets.token_hex(16)
    # Convert password and salt to bytes
    password_bytes = password.encode("utf-8")
    salt_bytes = salt.encode("utf-8")
    # Hash password
    password_hash = hashlib.pbkdf2_hmac(
        "sha256",
        password_bytes,
        salt_bytes,
        100000
    )
    # Convert hash to hexadecimal
    password_hash = password_hash.hex()
    return salt, password_hash
# ============================================================
# VERIFY PASSWORD
# ============================================================
def verify_password(password, salt, stored_hash):
    _, new_hash = hash_password(
        password,
        salt
    )
    return secrets.compare_digest(
        new_hash,
        stored_hash
    )
# ============================================================
# REGISTER
# ============================================================
def register():
    print("\n" + "=" * 50)
    print("USER REGISTRATION")
    print("=" * 50)
    username = input(
        "Enter username: "
    ).strip()
    # Empty username
    if not username:
        print(
            "\n❌ Username cannot be empty."
        )
        return
    # Check existing username
    if username in users:
        print(
            "\n❌ Username already exists."
        )
        return
    # Password
    password = getpass.getpass(
        "Enter password: "
    )
    confirm_password = getpass.getpass(
        "Confirm password: "
    )
    # Password match
    if password != confirm_password:
        print(
            "\n❌ Passwords do not match."
        )
        return
    # Password length
    if len(password) < 8:
        print(
            "\n❌ Password must contain "
            "at least 8 characters."
        )
        return
    # Hash password
    salt, password_hash = hash_password(
        password
    )
    # Store user
    users[username] = {
        "salt": salt,
        "password_hash": password_hash
    }
    print(
        "\n✅ Registration successful!"
    )
# ============================================================
# LOGIN
# ============================================================
def login():
    print("\n" + "=" * 50)
    print("LOGIN")
    print("=" * 50)
    username = input(
        "Username: "
    ).strip()
    if username not in users:
        print(
            "\n❌ Invalid username or password."
        )
        return None
    attempts = 3
    while attempts > 0:
        password = getpass.getpass(
            "Password: "
        )
        user = users[username]
        if verify_password(
            password,
            user["salt"],
            user["password_hash"]
        ):
            print(
                "\n✅ Login successful!"
            )
            print(
                f"Welcome, {username}!"
            )
            return username
        attempts -= 1
        print(
            f"\n❌ Incorrect password."
        )
        print(
            f"Attempts remaining: {attempts}"
        )
    print(
        "\n🔒 Too many failed attempts."
    )
    return None
# ============================================================
# PROFILE
# ============================================================
def profile(username):
    if username is None:
        print(
            "\n❌ Please login first."
        )
        return
    print("\n" + "=" * 50)
    print("PROFILE")
    print("=" * 50)
    print(
        "Username:",
        username
    )
    print(
        "Account Status: ACTIVE"
    )
# ============================================================
# LOGOUT
# ============================================================
def logout(username):
    if username is None:
        print(
            "\n❌ No user is currently logged in."
        )
        return None
    print(
        f"\n👋 Goodbye, {username}!"

    print(
        "You have been logged out."
    )
    return None
# ============================================================
# MAIN MENU
# ============================================================
def main():
    current_user = None
    while True:
        print("\n")
        print("=" * 50)
        print("🔐 LOGIN AUTHENTICATION SYSTEM")
        print("=" * 50)
        print("1. Register")
        print("2. Login")
        print("3. View Profile")
        print("4. Logout")
        print("5. Exit")
        print("=" * 50)
        choice = input(
            "Enter your choice: "
        )
        # ----------------------------------------------------
        # REGISTER
        # ----------------------------------------------------
        if choice == "1":
            register()
        # ----------------------------------------------------
        # LOGIN
        # ----------------------------------------------------
        elif choice == "2":
            if current_user is not None:
                print(
                    "\n⚠️ A user is already logged in."
                )
            else:
                current_user = login()
        # ----------------------------------------------------
        # PROFILE
        # ----------------------------------------------------
        elif choice == "3":
            profile(current_user)
        # ----------------------------------------------------
        # LOGOUT
        # ----------------------------------------------------
        elif choice == "4":
            current_user = logout(
                current_user
            )
        # ----------------------------------------------------
        # EXIT
        # ----------------------------------------------------
        elif choice == "5":
            print(
                "\n🔐 Thank you for using "
                "Login Authentication System."
            )
            break
        else:
            print(
                "\n❌ Invalid choice."
            )
# ============================================================
# PROGRAM START
# ============================================================
if __name__ == "__main__":
    main()
#
# ==================================================
#        🔐 LOGIN AUTHENTICATION SYSTEM
# ==================================================
# 1. Register
# 2. Login
# 3. View Profile
# 4. Logout
# 5. Exit
# ==================================================
# Enter your choice: 1
#
# ==================================================
#               USER REGISTRATION
# ==================================================
# Enter username: abhishek
# Enter password:
# Confirm password:
#
# ✅ Registration successful!
#
# ==================================================
#                  PROFILE
# ==================================================
# Username: abhishek
# Account Status: ACTIVE
#
# Username: abhishek
# Password:
# ❌ Incorrect password.
# Attempts remaining: 2
#
# Password:
# ❌ Incorrect password.
# Attempts remaining: 1
#
# Password:
# ❌ Incorrect password.
# Attempts remaining: 0
#
# 🔒 Too many failed attempts.
#
#                 START
#                   │
#                   ▼
#              Main Menu
#                   │
#        ┌──────────┼───────────┐
#        ▼          ▼           ▼
#    Register     Login       Exit
#        │          │
#        ▼          ▼
#  Generate Salt   Enter
#        │         Password
#        ▼           │
#    PBKDF2 Hash     ▼
#        │       Verify Hash
#        ▼           │
#    Store Hash      ▼
#                ┌───┴───┐
#                │       │
#              Valid   Invalid
#                │       │
#                ▼       ▼
#              Login   Retry
```
